```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("/content/sample_data/Social_Network_Ads .csv")
```

```python
# seperating input and target features:

x=df.iloc[:,2:4]
y=df.iloc[:,-1]
```

```python
# splitting the data into train and test:
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=45)
```

```python
from sklearn.svm import SVC
classifier=SVC(kernel="rbf")
classifier.fit(x_train,y_train)


# training score
y_train_pred=classifier.predict(x_train)
```

```python
# predicting the values for the test set
y_pred_test=classifier.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
print("Accuracy of the model for train data",accuracy_score(y_train_pred,y_train))
print("Accuracy of the model for test data",accuracy_score(y_test,y_pred_test))
```

- Parameter to perform hyperparameter tuning in SVM

1. Gamma : is something which we use for non- linear kernals such as RBF (Radial Basis Function) or polynomial to control the flecibility and complexity of the model. It control the spread of kernal influence.
- Gamma value (high)- decision bounday is complex. (overfitting)
- Gamma value (low)- very smooth.   (underfit)


2. Kernel - responsible for feature engineering.

3. C - parmeter which controls the trade off between achieving a wide decision margin and minimizing missclassification errors. It is a hyperparameter needed to be tuned for good model performance on unseen data.

```python
tuned_parameters=[{"kernel":["rbf"],"gamma":[np.random.rand(3)],"C":[1,10,100,20,1000]},
                  {"kernel":["linear"],"C":[1,10,100,20,1000]}]
```

```python
from sklearn.model_selection import GridSearchCV
clf= GridSearchCV(estimator=SVC(),
                  param_grid=tuned_parameters,
                  scoring="accuracy",
                  cv=5,
                  return_train_score=True,
                  verbose=1
                  )

clf.fit(x_train,y_train)
```
