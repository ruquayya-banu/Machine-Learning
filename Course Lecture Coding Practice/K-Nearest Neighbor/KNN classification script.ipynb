```python
# splitting the data into train and test

from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=56)

print(x_train.shape,x_test.shape)
print(y_train.shape,y_test.shape)
```

```python
# Encoding the gender column

from sklearn.preprocessing import OneHotEncoder

encoder=OneHotEncoder(drop="first",sparse_output=False)

one_hot_encoded=encoder.fit_transform(x_train[["Gender"]])
one_hot_encoded
x_train_trans=pd.DataFrame(one_hot_encoded, columns=encoder.get_feature_names_out(["Gender"]),index=x_train["Gender"].index)
x_train_trans
```

```python
one_hot_encoded=encoder.transform(x_test[["Gender"]])
one_hot_encoded
x_test_trans=pd.DataFrame(one_hot_encoded, columns=encoder.get_feature_names_out(["Gender"]),index=x_test["Gender"].index)
x_test_trans
```

```python
# merging the encoded feature into train and test data.

x_train["Gender"]=x_train_trans
x_train
x_test["Gender"]=x_test_trans
```

# Benefits of precision

- precision= TP/predicted yes
- High precision means model is good at correctly identifying positive cases and avoids incorrect labelling of negative cases as positive.
- A high precision spam filter will be very reliable in not marking important emails as spam, ensuring legitimate messages are not
  missed.

# Benefits of recall:

- recall= TP/Actual_yes
- high recall indicates that model is good at identifying all actual positive cases and not missing them.
- A high recall model will flag as many fraudlent transactions as possible minimizing the financial loss.

# Benefits of F1 score:
- F1= 2*precision*recall/(precision +recall)
- it is a measure of a model's prediction in ML calculated as the harmonic mean of precisiomn and recall.
- Particularly useful for imabalanced data.
- F1 score is valauble in applications like spam detection and fraud detection where FP and FN both can have significant consequences.
