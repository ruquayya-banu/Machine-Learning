```python
#categorie column
# this column is imbalanced because zero's are more then one's
df["Purchased"].value_counts()
```

```python
# splitting the input and target feature

# input feature- Gender Age EstimatedSalary	- columns	  
x = df.iloc[:,1:4]

# target feature - Purchased - column
y = df.iloc[:,4:]
```

```python
x
```

```python
y
```

```python
#splitting the data into train and test
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size= 0.3,random_state=56)
print(x_train.shape,x_test.shape)
print(y_train.shape,y_test.shape)
```

```python
# gender is a categoring column- it is nominal - so using one hot encoding technique 
```

```python
# Encoding the gender column - for trained data
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder(drop = "first", sparse_output = False) # drop will remove the last column so in output female category will not get
one_hot_encoded = encoder.fit_transform(x_train[["Gender"]]) #output it will give it in array format

# array Convert to DataFrame
x_train_trans = pd.DataFrame(
    one_hot_encoded,
    columns=encoder.get_feature_names_out(["Gender"]),
    index=x_train.index
)
x_train_trans
```

```python
# Encoding the gender column - for test data
one_hot_encoded = encoder.transform(x_test[["Gender"]]) #output it will give it in array format

# array Convert to DataFrame
x_test_trans = pd.DataFrame(
    one_hot_encoded,
    columns=encoder.get_feature_names_out(["Gender"]),
    index=x_test.index
)
x_test_trans
```

```python
# merging the encoded feature into train data
x_train["Gender"]= x_train_trans
x_train
```

```python
# merging the encoded feature into test data
x_test[["Gender"]] = x_test_trans
x_test
```

```python
# ******* above code is now made data ready to fit in your ML model **********
```

```python
# deciding the value of "k" 
# square root of count of training data points

import numpy as np
np.sqrt(x_train.shape[0]) #zero index
```

```python
# KNN Classification 
from sklearn.neighbors import KNeighborsClassifier
knn =  KNeighborsClassifier(n_neighbors= 17)
# fitting a data into the knn model
knn.fit(x_train, y_train)
```

```python
# prediciton 
y_pred = knn.predict(x_test)
y_pred
```

```python
#actual value 
# y_test.values  #in array format you will get output
y_test
```

```python
# finding accuracy of KNN model

from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

```python
# 79.16% accuracy in this model we got
```

```python
# error metrics in classification models are:-
# confusion matrix
# accuracy score
# precision
#recall
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(y_test,y_pred)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred))
```

```python
# benefits of precision
# precision = TP/predicted yes
# high precision means model is good at correctly identify positive cases and avoid incorrect labelling of negavtive cases as positive.
# a high precisin spam filter will be very reliable in not making important email as spam, ensuring legitimate message are n ot missed.
```

```python
#benifits of recall
# recall = TP/Actual yes
# high recall indicates that model is good at identifying all actual positive cases are not missing them.
# a high recall model will flag as many fraudulant transactions as possible minimizing the financial los.
```

```python
# benifits of F1 Score
# F1= 2*precision*recall/(precision +recall)
# it is a measure of a model's predicition in ML calculated as the harmonic mean of precision and recall.
# particularly useful for imbalanced data.
#F1 Score is valuable in application like spam and fraud dectection where FP and FN both can have significance consequences.
```
