```python
# column wise check for "?" value 
# in three column we have question mark(?) 

# frequency count of values in categorical columns only
for var in categorical:
 print((df[var].value_counts()/len(df))*100)
```

```python
#  methods for solving the "?" values in dataset
# "?" --->what percentage of values this "?" in it
# "<50(less than 50% values have quest mark in categorical columns in the dataset )" ---> replace it NAN then fill it with mode
# ">50 (greater than 50% values have quest mark in categorical columns in the dataset )"  ---> try to do backward or forward fill or drop it 
```

```python
# percent frequency distribution only for categorical columns to find "?" in percentage:
for var in categorical:
    print((df[var].value_counts()/len(df))*100)


# workclass column has 5% of "?" values
# occupation column has 5% of "?" values
# native_country column has 1% of "?" values
```

```python
# replacing the "?" with NAN
import numpy as np
# remove the white space before and after of the question mark
df["workclass"] = df["workclass"].str.strip()
# questiom mark replace with NaN
df["workclass"] = df["workclass"].replace("?",np.nan)
#check the question mark is filled with NaN or not
df["workclass"].value_counts()
```

```python
# remove the white space before and after of the question mark
df["occupation"] = df["occupation"].str.strip()
# questiom mark replace with NaN
df["occupation"] = df["occupation"].replace("?",np.nan)
#check the question mark is filled with NaN or not
df["occupation"].value_counts()
```

```python
# remove the white space before and after of the question mark
df["native_country"] = df["native_country"].str.strip()
# questiom mark replace with NaN
df["native_country"] = df["native_country"].replace("?",np.nan)
#check the question mark is filled with NaN or not
df["native_country"].value_counts()
```

```python
#check null entries in each column
df.isnull().sum()
```

```python
# imputing null values into categorical features

df.fillna({"workclass":df["workclass"].mode()[0]},inplace = True)
df.fillna({"occupation":df["occupation"].mode()[0]},inplace = True)
df.fillna({"native_country":df["native_country"].mode()[0]},inplace = True)

```

# Training the Model - Naive Bayes algorithm start

```python
# naive bayes algo are only for classification type- do not comes under regressor
# Naive bayses(NB) algorith are:-
# Gaussian NB
# multinominal NB
# Bernoulli NB
```

# Gaussian Naive Bayes classifier

```python
# naive bayes model training
```

```python
from sklearn.naive_bayes import GaussianNB

#instance of the class- inatantiate the model
gnb= GaussianNB()
gnb.fit(x_train,y_train)
```

```python
# predict the result
y_pred = gnb.predict(x_test)
y_pred
```

```python
# compare y_test(actual value) and y_pred(predicted value)
y_test
```

# Performance of the model 

```python
# checking the accuracy of the model
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
# 86.3% percent accuracy we got
```

```python
# creating confusion matrix
from sklearn.metrics import confusion_matrix,classification_report
confusion_matrix(y_test,y_pred)
# 833 and 765 are the wrong prediction
#6695 and 1475 are the correct prediction
```

```python
print(classification_report(y_test,y_pred))
# support column -  is total count of confusion matrix
```
