## Categorical features

```python
# finding all the categorical columns
categorical=[var for var in df.columns if df[var].dtype=="O"]

count=len(categorical)
print(f"There are {count} categorical variables")
print("The categorical variables are:\n",categorical)
```

```python
df[categorical].head()
```

```python
# frequency counts of values in categorical features
for var in categorical:
    print(df[var].value_counts())          # "?" ---> what percent of values have this ? in it
                                           # < 50% ---> replace it NAN then fill it with mode
                                           #>50% ---> try to do backward or forward fill or drop it.
```

```python
# percentage frequency distribution:

for var in categorical:
    print((df[var].value_counts()/len(df))*100)  #native_country,occupation,workclass (contains ?)
```

```python
# replacing the ? in the present columns with NAN
#df.replace({"workclass":np.NaN},inplace=True)
df["workclass"] = df["workclass"].str.strip()   # removing white spaces from front and end of the string.
df["workclass"]=df["workclass"].replace("?",np.nan)
```

```python
df["workclass"].value_counts()
```

```python
df["occupation"] = df["occupation"].str.strip()   # removing white spaces from front and end of the string.
df["occupation"]=df["occupation"].replace("?",np.nan)
```

```python
df["occupation"].value_counts()
```

```python
df["native_country"] = df["native_country"].str.strip()   # removing white spaces from front and end of the string.
df["native_country"]=df["native_country"].replace("?",np.nan)
df["native_country"].value_counts()
```

```python
# imputing the null values into categorical features
df.isnull().sum()
```

```python

df["workclass"].fillna(df["workclass"].mode()[0],inplace=True)

df["occupation"].fillna(df["occupation"].mode()[0],inplace=True)

df.fillna({"native_country":df["native_country"].mode()[0]},inplace=True)  # alternative way
```

```python
df.isnull().sum()
```

## Declaring input and target features:

```python
x= df.drop(["income"],axis=1)
y=df["income"]
```

## Splitting the data into train and test:

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=10)
```

```python
import category_encoders as ce
encoders=ce.OneHotEncoder(cols=['workclass', 'education', 'marital_status', 'occupation', 'relationship', 
                                'race', 'Gender', 'native_country'])
x_train_c=encoders.fit_transform(x_train)
x_test_c=encoders.transform(x_test)
```

## Feature Scaling:

```python
from sklearn.preprocessing import RobustScaler
scaler=RobustScaler()
x_train_n=scaler.fit_transform(x_train[numerical])  
x_train_n# answer will be in the format of array
```

```python
x_test_n=scaler.transform(x_test[numerical])  # answer will be in form of array
```

```python
#since x_train_n and x_test_n is in the form of array we cant consider their indexes for the DF conversion so we are taking x_train_c
#and x_test_c indexes reference over here

x_train_n=pd.DataFrame(x_train_n,columns=[numerical],index=x_train.index) 
x_train_n# dataFrame
```

```python
x_test_n=pd.DataFrame(x_test_n,columns=[numerical],index=x_test.index)
```

```python
x_train= pd.concat([x_train_c,x_train_n],axis=1)
x_test=pd.concat([x_test_c,x_test_n],axis=1)
```

```python
#training the model:
# Gaussian NB
# Multinomail NB
# Bernoulli NB
# some of the column names were in the format of tuple which model doesnot accept it only accept str column name or directly the values.
x_train=x_train.values
x_test=x_test.values

# training a Gaussian NB classifier on the training set

from sklearn.naive_bayes import GaussianNB

# inatantiate the model
gnb=GaussianNB()
gnb.fit(x_train,y_train)
```

```python
# predict the results

y_pred=gnb.predict(x_test)
y_pred
```

```python
y_test
```

```python
# checking the accuracy of the model
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

```python
# creating confusion matrix

from sklearn.metrics import confusion_matrix,classification_report
confusion_matrix(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```
