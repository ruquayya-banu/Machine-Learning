```python
# simple linear regression(SLR) we called when we have only two feature in dataset
```

```python
import pandas as pd
```

```python
df= pd.read_csv("C:/Ruquayya Banu/Data Science/Datasets/student_scores.csv")
```

```python
df
```

```python
# seperate input and target feature 
x= df[["Hours"]]
y= df[["Scores"]]
```

```python
# for linear regression we need to have dependency amongst x input feature and y target column
# scatter or heatplot we use to check the relationship of x and y
```

```python
import seaborn as sns
sns.scatterplot(data = df,x="Hours", y="Scores")
# x and y are in relationship
```

```python
# pearson's correlation coeffient 
sns.heatmap(df.corr(),annot = True)
# High positive coeffient correlation getting in output
```

```python
#splitting the data into train and test
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3,random_state = 56)
```

```python
# trainning the linear regression model
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(x_train,y_train)
```

```python
# intercept the best fit of line
regressor.intercept_
```

```python
# making the prediction - or  - test the model
y_pred= regressor.predict(x_test)
y_pred
```

```python
y_test
```

```python
# residual(error) analysis on training data
import matplotlib.pyplot as plt
plt.scatter(x_train["Hours"],y_train)# actuall value
plt.plot(x_train["Hours"],regressor.predict(x_train),color = "red") #best fit line it give
```

```python
# residual(error) analysis on test data
plt.scatter(x_test["Hours"],y_test)   # represent actual data point
plt.plot(x_test["Hours"],y_pred,color = "red") # represent predicted best fit line on test data
```

```python
# evaluation of SLR algorithm
```

```python
# mean absolute error
from sklearn import metrics
metrics.mean_absolute_error(y_test,y_pred)
```

```python
# mean squared error
metrics.mean_squared_error(y_test,y_pred)
```

```python
# root mean squared error
# to find a square root take the help of numpy
import numpy as np
np.sqrt(metrics.mean_squared_error(y_test,y_pred))
```
