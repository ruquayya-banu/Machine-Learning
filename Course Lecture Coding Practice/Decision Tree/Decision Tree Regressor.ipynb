```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import fetch_california_housing
california = fetch_california_housing(as_frame=True)
x = california.data
y=california.target
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=8)


from sklearn.tree import DecisionTreeRegressor
regressor=DecisionTreeRegressor()

#training the
regressor.fit(x_train,y_train)
```

```python
y_pred=regressor.predict(x_test)
```

```python
y_pred
```

```python
# R2---> regression score.
--> it is also called as coefficient of determination or goodness of fit.
```

```python
---> If I am changing the random state my accuracy of the model should not vary very far from the previous one which shows my data has
good variance
--> It checks how well the model fits the data by assesing the proportion of variance in the input variables.
---> The value of r2 ranges from 0-1. with 1 indicating the perfect fit for the model. 0 indicates that model does not explain any of the
varibility.
```

```python
from sklearn.metrics import r2_score
score=r2_score(y_pred,y_test)
```
