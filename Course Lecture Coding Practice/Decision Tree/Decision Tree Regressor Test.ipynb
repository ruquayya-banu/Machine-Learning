```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# loading dataset from sklearn 
from sklearn.datasets import fetch_california_housing
california  = fetch_california_housing(as_frame = True)
x = california.data
y = california.target
```

```python
california
```

```python
# splitting the data into train and test
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3,random_state =67)
```

```python
# Decision tree regressor 
from sklearn.tree import DecisionTreeRegressor
regressor = DecisionTreeRegressor()
```

```python
#train the Decision tree regressor model
regressor.fit(x_train,y_train)
```

```python
y_pred = regressor.predict(x_test)
```

```python
y_pred
```

```python
# R square or regression score or r2 --> accuracy measure
# it is also called as coefficient of determination or goodness of fit.
# if im changing the random state - my accuracy model should not be change  from the previous shows my data has good variance.
# it check how well the model fits the data by assesing the proportion of variance in the input variable.
# the value of R2 ranges from 0-1, with 1 indication the perfect fit for the model, where 0 indicates that model does not explain any of the varibility.
# only for regressor model R2 square 
```

```python
# calculating R2 Score
from sklearn.metrics import r2_score
score = r2_score(y_pred,y_test)
```

```python
score #57.8 % fit for the model - or accuracy of regressor model
```
